1. Principios Psicológicos del UX (Laws of UX)
Ley de Jakob: Los usuarios prefieren que los nuevos sitios funcionen de la misma manera que los que ya conocen
. Aprovechar sus modelos mentales preexistentes les permite centrarse en sus objetivos en lugar de aprender nuevas reglas
.
Ley de Fitts: El tiempo necesario para seleccionar un objetivo depende de su tamaño y de la distancia hasta él
. Las áreas táctiles deben ser suficientemente grandes, estar bien separadas y ubicarse en zonas de fácil acceso
.
Ley de Hick: El tiempo para tomar una decisión aumenta con el número y la complejidad de las opciones
. Reducir y simplificar las alternativas disponibles disminuye la carga cognitiva del usuario
.
Ley de Miller: La memoria de trabajo solo puede retener alrededor de 7 (± 2) elementos a la vez
. Organizar la información mediante agrupamiento (chunking) facilita su comprensión y memorización
.
Ley de Tesler y Umbral de Doherty: Todo sistema tiene una cantidad inherente de complejidad que el diseño debe gestionar en lugar de trasladársela al usuario
. Además, la interacción debe responder a un ritmo inferior a 400 ms para mantener la fluidez y productividad
.
2. Fundamentos de Diseño de Interfaz (Practical UI)
Reducción del costo de interacción: El diseño debe minimizar el esfuerzo físico y mental necesario para completar cualquier tarea dentro de la aplicación
.
Jerarquía visual y sistema de espaciado: Se debe ordenar la información variando el tamaño, peso y contraste del texto
. Aplicar un sistema de espaciado predefinido (en incrementos de 8 pt) ayuda a agrupar elementos relacionados y crea consistencia
.
Uso funcional del color: Es conveniente diseñar primero en blanco y negro para validar la estructura antes de incorporar color
. Utilizar una paleta de color acotada (reservando el color de marca para elementos interactivos) evita confundir al usuario
.
Consistencia y alineación: Mantener una alineación izquierda constante en textos e iconos facilita el recorrido visual
. La reutilización de componentes y patrones estándar acelera la toma de decisiones y reduce errores

1. Cambio hacia el Diseño Centrado en el Usuario
De la complejidad técnica a la simplicidad de uso: El diseño de experiencia de usuario (UX) prioriza resolver necesidades humanas sobre la simplicidad del código
. Se traslada la complejidad desde las personas hacia la tecnología; por ejemplo, la evolución de grifos de dos mandos a monomando requiere un mecanismo interno más elaborado, pero simplifica la interacción para el usuario
.
Alineación de objetivos: Proporcionar una experiencia fluida reduce la frustración del usuario y beneficia directamente a la organización (por ejemplo, facilitando procesos de registro o compra)
.
2. Fundamentos Biológicos y Psicológicos
Percepción y movimiento: La visión periférica humana está optimizada para detectar movimiento rápidamente
. Por esta razón, las notificaciones en apps se diseñan deslizándose desde el borde superior en lugar de aparecer mediante desvanecimiento
.
Psicología de la forma y el color: Las formas y colores evocan reacciones intuitivas
. Las figuras redondeadas (efecto Bouba) transmiten un tono más amigable y lúdico, mientras que los bordes rectos (Kiki) reflejan seriedad
. En cuanto al color, tonos como el azul favorecen la relajación, mientras que el rojo capta la atención inmediata
.
Leyes de Gestalt y Fitts: La ley de proximidad agrupa elementos cercanos en la mente del usuario
. La ley de Fitts establece que el tiempo para interactuar con un objetivo depende de su tamaño y distancia, haciendo imprescindible el uso de objetivos táctiles (touch targets) adecuados para los dedos
.
3. Reducción de Fricción y Modelos Mentales
Affordances y convenciones: Las affordances (elementos visuales que sugieren la acción de un objeto, como los tiradores de las puertas) deben ser claras para respetar el principio de menor asombro
. Iconos universales como la "X" para cerrar aprovechan convenciones aprendidas previamente por los usuarios
.
Modelo Mental vs. Modelo Conceptual: La interfaz debe comunicarse en el lenguaje del usuario, alineando la estructura de la app con su modelo mental preexistente
.
Intervención mínima: Es clave minimizar los pasos operativos ofreciendo entradas de datos flexibles, valores predeterminados inteligentes y manipulación directa en pantallas táctiles
.
4. Patrones de Diseño Móvil
Estructuras de navegación:
Menú inferior (Bottom menu): Se sitúa en la zona de alcance natural del pulgar, facilitando la interacción con una sola mano
.
Menú lateral/hamburguesa (Sliding drawer): Oculta opciones para liberar espacio en pantalla cuando existen múltiples secciones secundarias
.
Componentes y comportamientos: Uso del Botón de Acción Flotante (FAB) para la acción principal
, divulgación progresiva (progressive disclosure) para revelar funciones a medida que el usuario gana experiencia
 y gestos estandarizados como deslizar hacia abajo para actualizar (swipe down to refresh)
.
5. Investigación y Pruebas de Usabilidad
Herramientas de investigación: Creación de documentos de soporte como las Personas (ficticias pero representativas del usuario objetivo), Escenarios y User Journeys para mapear la interacción, pensamientos y emociones en cada paso
.
Evaluación con usuarios: Probar prototipos con usuarios reales es esencial para identificar puntos de fricción y corregir patrones problemáticos antes de la fase de desarrollo final
