1. Principios Psicológicos del UX (Laws of UX)
Ley de Jakob: Los usuarios prefieren que los nuevos sitios funcionen de la misma manera que los que ya conocen
. Aprovechar sus modelos mentales preexistentes les permite centrarse en sus objetivos en lugar de aprender nuevas reglas
.
Ley de Fitts: El tiempo necesario para seleccionar un objetivo depende de su tamaño y de la distancia hasta él
. Las áreas táctiles deben ser suficientemente grandes, estar bien separadas y ubicarse en zonas de fácil acceso
.
Ley de Hick: El tiempo para tomar una decisión aumenta con el número y la complejidad de las opciones
. Reducir y simplificar las alternativas disponibles disminuye la carga cognitiva del usuario
.
Ley de Miller: La memoria de trabajo solo puede retener alrededor de 7 (± 2) elementos a la vez
. Organizar la información mediante agrupamiento (chunking) facilita su comprensión y memorización
.
Ley de Tesler y Umbral de Doherty: Todo sistema tiene una cantidad inherente de complejidad que el diseño debe gestionar en lugar de trasladársela al usuario
. Además, la interacción debe responder a un ritmo inferior a 400 ms para mantener la fluidez y productividad
.
2. Fundamentos de Diseño de Interfaz (Practical UI)
Reducción del costo de interacción: El diseño debe minimizar el esfuerzo físico y mental necesario para completar cualquier tarea dentro de la aplicación
.
Jerarquía visual y sistema de espaciado: Se debe ordenar la información variando el tamaño, peso y contraste del texto
. Aplicar un sistema de espaciado predefinido (en incrementos de 8 pt) ayuda a agrupar elementos relacionados y crea consistencia
.
Uso funcional del color: Es conveniente diseñar primero en blanco y negro para validar la estructura antes de incorporar color
. Utilizar una paleta de color acotada (reservando el color de marca para elementos interactivos) evita confundir al usuario
.
Consistencia y alineación: Mantener una alineación izquierda constante en textos e iconos facilita el recorrido visual
. La reutilización de componentes y patrones estándar acelera la toma de decisiones y reduce errores

1. El espejismo de la caja de texto y el efecto ELIZA
La idea de que una caja de texto conversacional es la interfaz definitiva para la IA es un mito que oculta la complejidad probabilística del sistema
. Este diseño minimalista refuerza el efecto ELIZA, provocando que los usuarios proyecten inteligencia humana, empatía y razonamiento en motores meramente estadísticos
. La caja de texto suele hacer invisibles los errores y provoca que las personas traten a la IA como un buscador tradicional, pasando por alto la información contextual que el modelo realmente procesa
.
2. El marco de trabajo Entrada-Cómputo-Salida (Input-Computation-Output)
Entrada (Input): La intención del usuario viaja a través de tres canales simultáneos: el contexto implícito (documentos abiertos, historial o entorno), las instrucciones explícitas (prompts) y la manipulación directa (botones, selecciones y controles visuales)
.
Cómputo (Computation): Es la fase interna donde el modelo procesa tokens, enruta la solicitud y genera respuestas
. Dado que la espera transmite un mensaje sobre la confiabilidad del sistema, la latencia debe diseñarse deliberadamente como una superficie de interacción y retroalimentación
.
Salida (Output): Bajo la premisa de que "la salida no es la respuesta final", las respuestas deben presentarse como artefactos interactivos estructurados para ser claros, verificables, fundamentados (grounded), accionables y ajustables
.
3. Gestión de expectativas y fallos de la IA
Los usuarios tienden a asumir factualidad, neutralidad y conocimiento absoluto en los modelos
. A diferencia del software tradicional, los errores en la IA —como las alucinaciones y la sicofancia (la tendencia del modelo a darle la razón al usuario)— se entregan con fluidez gramatical y confianza algorítmica
. Por ello, el diseño de la interfaz debe incorporar fricciones intencionales que inviten a la verificación humana y eviten la sobreconfianza automatizada
.
4. Interfaces agénticas y el futuro multimodal
Para guiar a los sistemas agénticos que planifican y ejecutan tareas autónomas, la interfaz debe adoptar tres nuevos principios fundamentales
:
Revelar el plan (Reveal the Plan)
.
Priorizar lo más importante (Prioritize What Matters Most)
.
Diseñar para el control compartido (Design for Shared Control)
.
El futuro del diseño no se limita al chat, sino a la proliferación de modalidades que combinan conversaciones con canvas interactivos, herramientas de edición directa y niveles supervisados de automatización
