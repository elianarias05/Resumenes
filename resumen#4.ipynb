1. Principios Psicológicos del UX (Laws of UX)
Ley de Jakob: Los usuarios prefieren que los nuevos sitios funcionen de la misma manera que los que ya conocen
. Aprovechar sus modelos mentales preexistentes les permite centrarse en sus objetivos en lugar de aprender nuevas reglas
.
Ley de Fitts: El tiempo necesario para seleccionar un objetivo depende de su tamaño y de la distancia hasta él
. Las áreas táctiles deben ser suficientemente grandes, estar bien separadas y ubicarse en zonas de fácil acceso
.
Ley de Hick: El tiempo para tomar una decisión aumenta con el número y la complejidad de las opciones
. Reducir y simplificar las alternativas disponibles disminuye la carga cognitiva del usuario
.
Ley de Miller: La memoria de trabajo solo puede retener alrededor de 7 (± 2) elementos a la vez
. Organizar la información mediante agrupamiento (chunking) facilita su comprensión y memorización
.
Ley de Tesler y Umbral de Doherty: Todo sistema tiene una cantidad inherente de complejidad que el diseño debe gestionar en lugar de trasladársela al usuario
. Además, la interacción debe responder a un ritmo inferior a 400 ms para mantener la fluidez y productividad
.
2. Fundamentos de Diseño de Interfaz (Practical UI)
Reducción del costo de interacción: El diseño debe minimizar el esfuerzo físico y mental necesario para completar cualquier tarea dentro de la aplicación
.
Jerarquía visual y sistema de espaciado: Se debe ordenar la información variando el tamaño, peso y contraste del texto
. Aplicar un sistema de espaciado predefinido (en incrementos de 8 pt) ayuda a agrupar elementos relacionados y crea consistencia
.
Uso funcional del color: Es conveniente diseñar primero en blanco y negro para validar la estructura antes de incorporar color
. Utilizar una paleta de color acotada (reservando el color de marca para elementos interactivos) evita confundir al usuario
.
Consistencia y alineación: Mantener una alineación izquierda constante en textos e iconos facilita el recorrido visual
. La reutilización de componentes y patrones estándar acelera la toma de decisiones y reduce errores

1. La importancia del contexto
Antes de construir cualquier gráfico, se debe diferenciar entre el análisis exploratorio (comprender los datos por uno mismo) y el análisis explicativo (comunicar un mensaje específico a otros)
. El proceso inicia definiendo tres elementos clave:
Quién: La audiencia específica y la relación con el comunicador
.
Qué: La acción concreta que se busca desencadenar, junto con el mecanismo de entrega y el tono adecuado
.
Cómo: Los datos necesarios para estructurar la evidencia
. Para sintetizar el mensaje central, se recurre a técnicas como la historia de 3 minutos, la Gran Idea y el storyboarding (guion gráfico) para definir el flujo antes de crear diapositivas
.
2. Selección de visuales eficaces
Cada tipo de dato requiere una forma visual adecuada para evitar malentendidos
:
Texto simple: Recomendado cuando se comunica solo un número o cifra impactante
.
Tablas y mapas de calor: Útiles para audiencias diversas que necesitan buscar cifras específicas sin perder el contexto de color
.
Gráficos de puntos, líneas y barras: Ideales para tendencias temporales, comparaciones categóricas, variaciones acumuladas o diagramas de dispersión
.
Visuales a evitar: Se desaconseja rotundamente el uso de gráficos de sectores (pastel), donas, gráficos en 3D y ejes Y secundarios, ya que dificultan la comparación visual precisa y generan fricción cognitiva
.
3. Eliminación del desorden (Reducir la carga cognitiva)
El desorden visual incrementa la carga cognitiva de la audiencia
. Para optimizar la relación señal-ruido (o datos-tinta), se aplican los principios Gestalt de percepción visual, asegurando una alineación limpia, un uso intencional del espacio en blanco y la eliminación de elementos decorativos no esenciales
.
4. Enfocar la atención de la audiencia
El cerebro procesa información visual a través de la memoria icónica y a corto plazo
. Mediante el uso estratégico de atributos preatentivos —tales como el tamaño, la posición en la página y el uso controlado/consistente del color— se dirige la mirada del espectador directamente a los datos más importantes
.
5. Pensar como un diseñador y estructurar la narrativa
Un diseño efectivo incorpora asequibilidad (affordances), accesibilidad y estética para generar empatía y aceptación en el usuario
. Finalmente, la presentación se organiza en una estructura narrativa con inicio, medio y fin, apoyada en tácticas de claridad y el poder de la repetición para guiar a la audiencia desde la evidencia hasta la toma de decisiones
