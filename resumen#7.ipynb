1. Principios Psicológicos del UX (Laws of UX)
Ley de Jakob: Los usuarios prefieren que los nuevos sitios funcionen de la misma manera que los que ya conocen
. Aprovechar sus modelos mentales preexistentes les permite centrarse en sus objetivos en lugar de aprender nuevas reglas
.
Ley de Fitts: El tiempo necesario para seleccionar un objetivo depende de su tamaño y de la distancia hasta él
. Las áreas táctiles deben ser suficientemente grandes, estar bien separadas y ubicarse en zonas de fácil acceso
.
Ley de Hick: El tiempo para tomar una decisión aumenta con el número y la complejidad de las opciones
. Reducir y simplificar las alternativas disponibles disminuye la carga cognitiva del usuario
.
Ley de Miller: La memoria de trabajo solo puede retener alrededor de 7 (± 2) elementos a la vez
. Organizar la información mediante agrupamiento (chunking) facilita su comprensión y memorización
.
Ley de Tesler y Umbral de Doherty: Todo sistema tiene una cantidad inherente de complejidad que el diseño debe gestionar en lugar de trasladársela al usuario
. Además, la interacción debe responder a un ritmo inferior a 400 ms para mantener la fluidez y productividad
.
2. Fundamentos de Diseño de Interfaz (Practical UI)
Reducción del costo de interacción: El diseño debe minimizar el esfuerzo físico y mental necesario para completar cualquier tarea dentro de la aplicación
.
Jerarquía visual y sistema de espaciado: Se debe ordenar la información variando el tamaño, peso y contraste del texto
. Aplicar un sistema de espaciado predefinido (en incrementos de 8 pt) ayuda a agrupar elementos relacionados y crea consistencia
.
Uso funcional del color: Es conveniente diseñar primero en blanco y negro para validar la estructura antes de incorporar color
. Utilizar una paleta de color acotada (reservando el color de marca para elementos interactivos) evita confundir al usuario
.
Consistencia y alineación: Mantener una alineación izquierda constante en textos e iconos facilita el recorrido visual
. La reutilización de componentes y patrones estándar acelera la toma de decisiones y reduce errores

1. Definición de "Móvil" y Enfoque Universal
Más allá del teléfono inteligente: Los autores definen lo "móvil" no solo como smartphones, sino como cualquier sistema portátil y contextual (reproductores, e-readers, GPS, cámaras, kioscos) que interactúa con el usuario fuera de las dinámicas tradicionales del escritorio
.
Patrones universales: Ofrece soluciones de diseño independientes del sistema operativo o plataforma, centrándose en la arquitectura de información, el diseño visual y la interacción humana
.
2. Principios Fundamentales del Diseño Móvil
La vida del usuario es lo primero: Los dispositivos móviles se utilizan en breves ráfagas de atención e interrupciones del entorno real; las interfaces no deben interrumpir la vida del usuario salvo que sea imprescindible
.
Uso activo de sensores e inteligencia: Las aplicaciones deben aprovechar el contexto y los sensores del hardware (GPS, acelerómetros, calendarios) para anticipar necesidades y actuar proactivamente
.
Protección de los datos del usuario: Dado que escribir en pantallas pequeñas propicia errores, el sistema debe incluir salvaguardas implícitas y explícitas (Cancel Protection, Exit Guard) para evitar la pérdida accidental de información
.
3. Arquitectura y Composición de Páginas (El Wrapper)
El marco contenedor (Wrapper): Es la plantilla modular o marco consistente que rodea el contenido principal en cada pantalla (menús, títulos, barra de notificaciones)
. Su uso predecible reduce la carga cognitiva y acelera el aprendizaje del usuario
.
Patrones de composición clave:
Scroll: Desplazamiento fluido preferentemente en un solo eje, evitando manipular barras diminutas e impulsando la interacción gestual directa
.
Menús fijos y revelables: Organización de opciones frecuentes u opcionales sin saturar el área de visualización (viewport)
.
Barra de estado (Annunciator Row) y Notificaciones: Zonas dedicadas a indicadores del sistema y alertas claras con descripciones legibles
.
4. Componentes y Visualización de Información
Modelos mentales y psicología Gestalt: La organización visual sigue el modelo de interacción de Donald Norman (affordances, mapeo, feedback, restricciones) y los principios de la Gestalt para facilitar la interpretación de datos
.
Formatos de visualización: Patrones como Vertical List, Grid, Film Strip y Slideshow permiten estructurar grandes volúmenes de información optimizando el valioso espacio en pantalla (screen real estate)
.
Revelación progresiva: Patrones como Windowshade y Pop-Up permiten desplegar detalles adicionales o paneles de control sin perder el contexto de la página principal
.
5. Navegación, Widgets y Control Gestual
Acceso lateral y Drilldown: Widgets para navegar entre secciones del mismo nivel jerárquico (Tabs, Peel Away, efectos 3D simulados)
 y elementos de profundización jerárquica (Link, Button, Icon, Stack of Items), recomendando no superar los 2 o 3 niveles de profundidad para evitar confusiones
.
Controles gestuales y sensoriales:
On-screen Gestures: Toques, arrastres o pellizcos (pinch to zoom) que simulan la manipulación de objetos físicos reales
.
Kinesthetic Gestures: Respuestas a movimientos del propio dispositivo (como agitar para borrar o colocar pantalla abajo para silenciar)
.
Retroalimentación inmediata (Feedback): Cualquier acción del usuario debe generar una respuesta inmediata (visual, auditiva o háptica/vibración) para confirmar que el comando fue procesado correctamente
